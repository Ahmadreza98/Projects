# How to use Seaborn?

## Part - 5

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
student_performance_df = pd.read_csv("DATA/StudentsPerformance.csv")
```

```python
student_performance_df.head()
```

```python
sns.jointplot(data=student_performance_df, x='math score', y='reading score')
```

```python
sns.jointplot(data=student_performance_df, x='math score', y='reading score', kind='hex')
```

```python
sns.jointplot(data=student_performance_df, x='math score', y='reading score', hue='gender')
```

```python
sns.pairplot(data=student_performance_df, hue='gender')
```

```python
sns.pairplot(data=student_performance_df, hue='gender', corner=True)
```
