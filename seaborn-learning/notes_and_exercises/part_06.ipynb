```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
student_performance_df = pd.read_csv("DATA/StudentsPerformance.csv")
```

```python
student_performance_df.head()
```

The **catplot** function is popular in the **seaborn**.

```python
sns.catplot(data= student_performance_df, x='gender', y='math score')
```

```python
plt.figure(figsize=(12,4), dpi=300)
sns.catplot(data= student_performance_df, x='gender', y='math score', row='lunch', hue='gender', palette='Set2', col='race/ethnicity')
```

```python
sns.catplot(data= student_performance_df, x='gender', y='math score', row='race/ethnicity', hue='gender', palette='Set2', col='lunch')
```

We can change the input value of **kind** in **catplot** function, change the curve plot.

```python
sns.catplot(data= student_performance_df, x='gender', y='math score', row='race/ethnicity', hue='gender', palette='Set2', col='lunch', kind='bar')
```
