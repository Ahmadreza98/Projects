# How to use Seaborn?

## Part - 1

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
office_df = pd.read_csv('DATA/dm_office_sales.csv')
```

```python
office_df.head()
```

Now, We want to plot the salary column based on sales with **Seaborn** and show different to **Matplotlib**.

```python
plt.scatter(office_df['salary'], office_df['sales'])
```

```python
sns.scatterplot(x='salary', y='sales', data=office_df)
```

For change figure-size we should be setup with **Matplotlib**.

```python
plt.figure(figsize=(10, 6), dpi=200)
sns.scatterplot(x='salary', y='sales', data=office_df)
```

one of the input value of the **Scatterplot** function in the **Seaborn** is the **hue**. The feature of that, categorizing data (categorical or numeric) based on that.

```python
plt.figure(figsize=(12, 4), dpi=200)
sns.scatterplot(x='salary', y='sales', data=office_df, hue='level of education')
```

```python
plt.figure(figsize=(12, 4), dpi=200)
sns.scatterplot(x='salary', y='sales', data=office_df, hue='salary')
```

Also, we can add feature **size** to data.

```python
plt.figure(figsize=(12, 4), dpi=200)
sns.scatterplot(x='salary', y='sales', data=office_df, hue='salary', size='salary', sizes=(20, 200))
```

Finally, we can change transparent with input value **alpha**.

```python
plt.figure(figsize=(12, 4), dpi=200)
sns.scatterplot(x='salary', y='sales', data=office_df, hue='salary', alpha=0.5, size='salary', sizes=(20, 200))
```

With the input value **Style**:

```python
plt.figure(figsize=(12, 4), dpi=200)
sns.scatterplot(x='salary', y='sales', data=office_df, hue='salary', alpha=0.5, size='salary', sizes=(20, 200), style='level of education')
```

Also, we can save figure with formats of jpg, png, svg and etc.

```python
plt.figure(figsize=(12, 4), dpi=200)
sns.scatterplot(x='salary', y='sales', data=office_df, hue='salary', alpha=0.5, size='salary', sizes=(20, 200), style='level of education')
plt.savefig('Images/office_sales_scatterplot.png', dpi=200, bbox_inches='tight')
```
