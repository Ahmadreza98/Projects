# How to use Seaborn?

## Part - 3

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
office_df = pd.read_csv('DATA/dm_office_sales.csv')
```

```python
office_df.head()
```

In this part, we want working to categorical data.

```python
plt.figure(figsize=(10,4))
sns.countplot(data=office_df, x='level of education')
```

```python
plt.figure(figsize=(10,4))
sns.countplot(data=office_df, x='division', hue='division', palette='deep')
```

```python
plt.figure(figsize=(10,4))
sns.countplot(data=office_df, x='division', hue='level of education', palette='deep')
```

```python
plt.figure(figsize=(12,4))
sns.barplot(data=office_df, x='level of education', y='salary', hue='level of education', palette='deep')
```

```python
plt.figure(figsize=(12,4), dpi=300)
sns.barplot(data=office_df, x='level of education', y='salary', hue='division', palette='deep', estimator=np.mean, ci='sd')
plt.legend(loc=(1.01,0.3))
```
