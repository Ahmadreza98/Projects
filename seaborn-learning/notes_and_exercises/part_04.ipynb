# How to use Seaborn?

## Part - 5

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
student_performance_df = pd.read_csv("DATA/StudentsPerformance.csv")
```

```python
student_performance_df.head()
```

```python
sns.boxplot(data=student_performance_df, x='math score')
```

```python
sns.boxplot(data=student_performance_df, x='math score', y='test preparation course', hue='test preparation course')
```

```python
sns.boxplot(data=student_performance_df, x='math score', y='parental level of education', hue='test preparation course', palette='Set3')

plt.legend(loc=(1.01,0.5))
```

```python
plt.figure(figsize=(12,4), dpi=200)
sns.boxplot(data=student_performance_df, x='parental level of education', y='math score', hue='test preparation course', palette='Set3')
plt.legend(loc=(1.01,0.5))
```

```python
plt.figure(figsize=(12,4), dpi=200)
sns.violinplot(data=student_performance_df, x='parental level of education', y='math score', hue='test preparation course', palette='Set3')
plt.legend(loc=(1.01,0.5))
```

```python
plt.figure(figsize=(12,4), dpi=200)
sns.violinplot(data=student_performance_df, x='parental level of education', y='math score', hue='test preparation course', palette='Set3', split=True)
plt.legend(loc=(1.01,0.5))
```

```python
plt.figure(figsize=(12,4), dpi=200)
sns.swarmplot(data=student_performance_df, x='math score', y='gender', size=3, hue='gender')
```
