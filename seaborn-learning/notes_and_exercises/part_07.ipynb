```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
countries_df.set_index('Countries', inplace=True)
```

```python
sns.heatmap(data= countries_df)
```

```python
sns.heatmap(data= countries_df.drop('Life expectancy', axis=1))
```

```python
sns.heatmap(data= countries_df.drop('Life expectancy', axis=1), annot=True, linewidths=0.1)
```
