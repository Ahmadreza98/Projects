# How to use Seaborn?

## Part - 2

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
office_df = pd.read_csv('DATA/dm_office_sales.csv')
```

```python
office_df.head()
```

This session explain to about **distribution** plots.

```python
plt.figure(figsize=(12, 4))
sns.rugplot(x='salary', data=office_df, height=0.5)
```

Display of data distribution in the figure above.

```python
sns.displot(x='salary', data=office_df, bins=30, kde=True, rug=True, height=4, aspect=3)
```

```python
sns.kdeplot(x='salary', data=office_df, fill=True, alpha=0.5, linewidth=2, color='blue')
```
