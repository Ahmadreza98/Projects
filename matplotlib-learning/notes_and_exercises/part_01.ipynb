# How to use Matplotlib?
## Part - 1

One of the most popular libraries in Python, **Matplotlib**. That using for visualization to show the **Data** on a figure.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
years = np.arange(2010,2026)
temperature_of_year = [0.65,1.02,0.25,0.45,0.16,0.88,0.75,1.2,1.1,0.9,0.85,0.65,0.85,0.54,0.23,0.12]
```

```python
years
```

Now, to show a simple plot in a figure

```python
plt.plot(years,temperature_of_year)
```

But, this figure is not good, because that exist any label for x-axis, y-axis, title and etc.

For add this properties to figure as follow:

```python
plt.plot(years,temperature_of_year)
plt.xlabel('Year')
plt.ylabel('Rate of temperature')
plt.title('Temperature rate changes by year')
plt.grid()
```

```python
Customer_1 = [800, 538, 224, 531, 492, 214, 490, 645, 412, 671, 748, 307]
Customer_2 = [340, 395, 483, 716, 411, 383, 133, 583, 670, 241, 495, 708]
Customer_3 = [429, 280, 541, 688, 729, 100, 236, 641, 382, 705, 204, 305]

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
```

```python
plt.plot(months, Customer_1, 'o-', label='Customer_1', color='red')
plt.plot(months, Customer_2, '^-', label='Customer_2', color='green')
plt.plot(months, Customer_3, 'd-', label='Customer_3', color='blue')
plt.xlabel('Months')
plt.ylabel('K.W.H (Kilo-Watt-Hour)')
plt.title('Building Electricity Consumers')
plt.legend()
plt.show()
```

We can use `subplot` function for figure management.

```python
plt.subplot(2,1,1)
plt.plot(months, Customer_1, 'o-', label='Customer_1', color='red')
plt.subplot(2,1,2)
plt.plot(months, Customer_3, 'd-', label='Customer_3', color='blue')
plt.show()
```
