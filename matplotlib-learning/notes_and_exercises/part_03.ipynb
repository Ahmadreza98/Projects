# How to use Matplotlib
## Part - 3

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Customer_1 = [800, 538, 224, 531, 492, 214, 490, 645, 412, 671, 748, 307]
Customer_2 = [340, 395, 483, 716, 411, 383, 133, 583, 670, 241, 495, 708]
Customer_3 = [429, 280, 541, 688, 729, 100, 236, 641, 382, 705, 204, 305]

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
```

To show a figure of bar-chart as follow:

```python
plt.bar(months, Customer_1)
plt.bar(months, Customer_2)
plt.xlabel('Months')
plt.ylabel('Consume of Customer (KWh)')
plt.show()
```

If we want to show two bar-chart to side by side as follow:

```python
months_bar = np.arange(12)
bar_width = 0.4

plt.bar(months_bar, Customer_1, width=bar_width, color='blue', label='customer_1')
plt.bar(months_bar-bar_width, Customer_2,  width=bar_width, color='red', label='customer_2')
plt.xlabel('Months')
plt.ylabel('Consume of Customer (KWh)')
plt.xticks(months_bar-bar_width/2, set(months))
plt.legend()
plt.show()
```
