```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

For to create several plots on a figure we can use **subplot** function.

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
axes[0].plot([1,2,3],[1,2,3])
axes[1].plot([1,2,3],[1,2,3])
```

In the above figure, there are two columns and one row. So to access that, we can use `axes[0]` or `axes[1]`.

```python
fig, axes = plt.subplots(nrows=2, ncols=2)
axes[0][0].plot([1,2,3],[1,2,3])
axes[1][1].plot([1,2,3],[1,2,3])
axes[0][1].plot([1,-2,3],[1,2,3])
axes[1][0].plot([1,-2,3],[1,2,3])
```

In the above figure, there are two columns and two row. That is same a 2x2 matrix. So to access that, we can use `axes[0][0]`, `axes[1][1]`, `axes[0][1]`,`axes[1][0]`.

```python
fig, axes = plt.subplots(nrows=2, ncols=2)

x = np.arange(-10,10)
k = 2

for i in np.arange(2):
    for j in np.arange(2):
        axes[i][j].plot(x, x**k)
        k = k + 1
```

Exist a problem in above figure, x-axes and y-axes, for solve this problem we can use **tight_layout** function.

```python
fig, axes = plt.subplots(nrows=2, ncols=2)

x = np.arange(-10,10)
k = 2

for i in np.arange(2):
    for j in np.arange(2):
        axes[i][j].plot(x, x**k)
        axes[i][j].set_title(f'Degree {k}')
        k = k + 1
plt.suptitle("Draw Different of digree power")
plt.tight_layout()
```
