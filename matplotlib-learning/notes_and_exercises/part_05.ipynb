```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Now, we want use **Advance Write Code** to show a figures.

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot([1,2,3,4],[70,90, 150, 280])
```

We can add a new axes based on nested axes in a figure.

```python
fig = plt.figure()
ax_1 = fig.add_axes([0,0,1,1])
ax_1.plot([1,2,3,4],[70,90, 150, 280])

ax_2 = fig.add_axes([0.2,0.4,0.4,0.4])
ax_2.scatter([1,2,3,4],[70,90,150,280])
```

Also, we can add label, title, legend etc, for each axes.

```python
fig = plt.figure()
ax_1 = fig.add_axes([0,0,1,1])
ax_1.plot([1,2,3,4],[70,90, 150, 280])
ax_1.set_xlabel('Number')
ax_1.set_ylabel('Price')
ax_1.set_title("Consume of Customer")
ax_2 = fig.add_axes([0.2,0.4,0.4,0.4])
ax_2.scatter([1,2,3,4],[70,90,150,280])
ax_2.set_xlabel('Number')
ax_2.set_ylabel('Price')
ax_2.set_title("Consume of Customer")
```

```python
fig = plt.figure(figsize=(10,6), dpi=300)
ax_1 = fig.add_axes([0,0,1,1])
ax_1.plot([1,2,3,4],[70,90, 150, 280])
ax_1.set_xlabel('Number')
ax_1.set_ylabel('Price')
ax_1.set_title("Consume of Customer")
ax_2 = fig.add_axes([1,1,0.4,0.4])
ax_2.scatter([1,2,3,4],[70,90,150,280])
ax_2.set_xlabel('Number')
ax_2.set_ylabel('Price')
ax_2.set_title("Consume of Customer")
```
