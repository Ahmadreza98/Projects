```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

We can determining **the location of legends** of the figure.

```python
fig, axes = plt.subplots(nrows=2, ncols=2)

x = np.arange(-10,10)
k = 2

for i in np.arange(2):
    for j in np.arange(2):
        axes[i][j].plot(x, x**k, label=f'clip{i}')
        axes[i][j].set_title(f'Degree {k}')
        axes[i][j].legend(loc=(1,1))
        k = k + 1
plt.suptitle("Draw Different of digree power")
plt.tight_layout()
```

We can add **the marker** for each plot.

```python
fig = plt.figure()
axes = fig.add_axes([0,0,1,1])
Customer_1 = [800, 538, 224, 531, 492, 214, 490, 645, 412, 671, 748, 307]
Customer_2 = [340, 395, 483, 716, 411, 383, 133, 583, 670, 241, 495, 708]
Customer_3 = [429, 280, 541, 688, 729, 100, 236, 641, 382, 705, 204, 305]

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
axes.plot(months, Customer_1, marker='^')
axes.plot(months, Customer_2, marker='o')
axes.plot(months, Customer_3, marker='*')
```

Also, we can define **the line-width** and **the line-style** of plot.

```python
fig = plt.figure()
axes = fig.add_axes([0,0,1,1])
Customer_1 = [800, 538, 224, 531, 492, 214, 490, 645, 412, 671, 748, 307]
Customer_2 = [340, 395, 483, 716, 411, 383, 133, 583, 670, 241, 495, 708]

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
axes.plot(months, Customer_1, linewidth=3)
axes.plot(months, Customer_2, linestyle='--')
```

More setup for marker of plot.

```python
fig = plt.figure()
axes = fig.add_axes([0,0,1,1])
Customer_1 = [800, 538, 224, 531, 492]
Customer_2 = [340, 395, 483, 716, 495]

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
axes.plot(months, Customer_1, linewidth=3)
axes.plot(months, Customer_2, linestyle='--', markersize=10, marker='^', markerfacecolor='red', markeredgewidth=3, markeredgecolor='black')
```
