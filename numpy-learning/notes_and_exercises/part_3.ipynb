# HOW TO USE NUMPY?
## Part - 3

```python
import numpy as np
```

```python
np1 = np.array([[1,2],[3,4]])
np2 = np.array([[5,6],[7,8]])
```

**Note**: Performing mathematical operations on arrays (matrices) requires that they be the same dimensions.

**SUM** operation arrays and numbers.

**array** + **array**:

```python
print(np1+np2)
```

**array** + **number**:

```python
print(np1+5)
```

SUM with **FUNCTION**

```python
print(np.add(np1,np2))
```

**MINUS** operation arrays and numbers

**array** - **array**:

```python
print(np1 - np2)
```

**array** - **number**:

```python
print(np1 - 5)
```

MINUS with **FUNCTION**

```python
print(np.subtract(np1,np2))
```

**MULTIPLE** operation arrays and number

- **Matrix Multiple**

```python
print(np1@np2)
```

- **Cells Multiple**

```python
print(np1*np2)
```

```python
print(np1 * 5)
```

**DIVIDE** operation arrays and number

```python
print(np1 / np2) # With the point number
```

```python
print(np2 // np1) # Without the point number (Just Integer part)
```
