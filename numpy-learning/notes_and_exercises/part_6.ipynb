# HOW TO USE NUMPY?
## Part - 6

```python
import numpy as np
```

```python
ar = np.random.randint(1,100,(20,5))
```

```python
print(ar)
```

**Extract** a part of array

```python
print(ar[0:3])
```

```python
print(ar[:,0:3])
```

```python
print(ar[0:5,0:3])
```

```python
B = ar[0:2,2:5]
```

```python
print(B)
```

In fact, the alias variable **B** is part of the entire array **ar** and by changing its values, changes in **ar** are also taken into account.

```python
B[:] = 100
```

```python
print(ar)
```

To solve the problem, The **ar** array should be a copy to **B**

```python
B = ar.copy()
```

Conditional orders can be applied directly to arrays, the output is an array of true and false.

```python
print(B > 50)
```

And show value that array

```python
print(B[B>50])
```

Also, we can use several conditions to find our result.

```python
print(B[(B>50) & (B<80)])
```
