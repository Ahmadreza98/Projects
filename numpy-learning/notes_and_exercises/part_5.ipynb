# HOW TO USE NUMPY?
## Part - 5

```python
import numpy as np
```

```python
ar = np.random.randint(1,100,(5,5))
print(ar)
```

Operation **statistics** in **Array**

**Mean** in array:

```python
print(np.mean(ar))
```

```python
print(np.mean(ar, axis=0))
```

```python
print(np.mean(ar,axis=1))
```

**median** in array:

```python
print(np.median(ar))
```

```python
print(np.median(ar, axis=0))
```

```python
print(np.median(ar, axis=1))
```

**Variants** in array:

```python
print(np.var(ar))
```

```python
print(np.var(ar, axis=0))
```

```python
print(np.var(ar, axis=1))
```

**Standard deviation** in array:

```python
print(np.std(ar,axis=0))
```

```python
print(np.std(ar,axis=1))
```

To find a **maximum** or **minimum** in array:

```python
print(np.max(ar))
```

```python
print(np.max(ar,axis=0))
```

```python
print(np.max(ar,axis=1))
```

```python
print(np.min(ar))
```

```python
print(np.min(ar,axis=0))
```

```python
print(np.min(ar,axis=1))
```

To find argument of **maximum** and **minimum** in array

```python
print(np.argmax(ar))
```

```python
print(np.argmin(ar))
```

To find **dimension** of array

```python
print(np.shape(ar))
```

To **reshape** in array

```python
ar2 = np.linspace(0,100,30)
print(ar2)
```

```python
print(np.reshape(ar2,(30,1)))
```

```python
print(np.reshape(ar2,(5,6)))
```
