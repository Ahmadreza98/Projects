# HOW TO USE NUMPY?
## Part - 2

```python
import numpy as np
```

Create 100 numbers in a specified range

```python
x = np.linspace(10,20,100)
print(x)
```

Create a array with **random** data with **standard normal distribution**

```python
x = np.random.standard_normal((4,4))
print(x)
```

Create a array with **random** data with **normal distribution**

```python
x = np.random.normal(3,0.25,(3,3))
print(x)
```

Create a array with **random** data with **uniform**

```python
x = np.random.uniform(3,2,(3,3))
print(x)
```

Create a array with **random** data with **rand** between 0 and 1.

```python
x = np.random.rand(3,3)
print(x)
```

Create a array with integer **random** data with **randint**.

```python
x = np.random.randint(3,10,(4,4))
print(x)
```

If we want create two random data same with use **seed**

```python
np.random.seed(43)

x = np.random.randint(10, 20)
print(f"First: {x}")

np.random.seed(43)

x = np.random.randint(10, 20)
print(f"Second: {x}")
```
