```python
import numpy as np
import pandas as pd
```

```python
tips_df = pd.read_csv('DATA/tips.csv')
```

```python
tips_df.head()
```

```python
tips_df.set_index('Payment ID', inplace=True)
```

We can **Sort** **DataFrame** based on several **Features**

```python
tips_df.sort_values(['tip', 'size'], ascending=True)
```

Find **Maximum** value and its index for each feature.

```python
tips_df['tip'].max()
```

```python
tips_df['tip'].idxmax()
```

Find **Minimum** value and its index for each feature.

```python
tips_df['tip'].min()
```

```python
tips_df['tip'].idxmin()
```

```python
tips_df.loc[f'{tips_df['total_bill'].idxmax()}']
```

Display the extent of **Direct** or **Inverse** relationships of features within a **DataFrame**

```python
tips_df.corr(numeric_only=True)
```

**Count** the values for each feature in **DataFrame**

```python
tips_df['sex'].value_counts()
```

```python
tips_df['day'].unique()
```

To find **Duplicated** rows

```python
tips_df.duplicated()
```

Drop **Duplicated** rows

```python
tips_df.drop_duplicates()
```

For conditional order and find to values between two values

```python
tips_df[tips_df['total_bill'].between(15,20)]
```

To show `n` largest in feature of DataFrame

```python
tips_df['tip'].nlargest(10)
```

To show `n` smallest in feature of DataFrame

```python
tips_df['tip'].nsmallest(10)
```

Create a random sample of DataFrame

```python
tips_df.sample(frac=0.1)
```
