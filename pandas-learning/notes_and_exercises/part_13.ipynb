```python
import numpy as np
import pandas as pd
```

**Working** with **String** in **Pandas**

```python
my_str = pd.Series(['Ali','Mohammad','Reza',"Hadi"])
```

To **Uppercase** text

```python
my_str.str.upper()
```

To **Lowercase** text

```python
my_str.str.lower()
```

Check is **Digit**

```python
my_str.str.isdigit()
```

Check is **Alphabet**

```python
my_str.str.isalpha()
```

Check is **Alph/Num**

```python
my_str.str.isalnum()
```
