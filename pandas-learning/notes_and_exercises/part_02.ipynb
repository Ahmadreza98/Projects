The **Mathematics Operation** in **Pandas** for **Series**

**SUM** operation in **Series**

```python
print(shopping_of_provinces_month_1+shopping_of_provinces_month_2)
```

The reason for **NaN** appearing in the addition of two series is the absence of identical keys; to solve this problem, the following method is used.

```python
total_shopping_month_1_and_2 = shopping_of_provinces_month_1.add(shopping_of_provinces_month_2, fill_value=0)
print(total_shopping_month_1_and_2)
```

```python
total_shopping_month_1_and_2 = shopping_of_provinces_month_1.subtract(shopping_of_provinces_month_2, fill_value=0)
print(total_shopping_month_1_and_2)
```
