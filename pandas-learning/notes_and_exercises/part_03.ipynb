```python
import pandas as pd
```

```python
my_data = pd.Series([1000,2300,4200,1500], index=['T', 'M', 'F', 'A'])
```

To Exist **2** method to access of data in **Series**.

```python
print(my_data['M'])
```

```python
print(my_data.M)
```

Now, exist **2** method others for find data in **Series**

```python
print(my_data.loc['M'])
```

```python
print(my_data.iloc[0])
```

```python
my_data[0:1]
```

Some indexes require the inclusion of dates. For this reason, the **Pandas library** has a **attribute** called **Datetime** that performs this task for us.

```python
dates_1 = pd.date_range("20260510", periods=4)
```

```python
dates_1
```

```python
my_data.index = dates_1
```

```python
my_data
```
