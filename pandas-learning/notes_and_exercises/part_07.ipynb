```python
import numpy as np
import pandas as pd
```

Set `tips.csv` DataFrame

```python
tips_df = pd.read_csv('DATA/tips.csv')
```

```python
tips_df.head()
```

```python
tips_df['Total_pay'] = tips_df['total_bill'] + tips_df['tip']
```

To change **Index** in DataFrame to other features. 

```python
tips_df.set_index('Payment ID', inplace=True)
```

Now, we can search in DataFrame with new index.

```python
tips_df.loc["Sun2251"]
```

```python
tips_df.iloc[10]
```

Conditional Operation in DataFrame

```python
tips_df[tips_df['total_bill'] > 15].head()
```

```python
tips_female_total_pay_df = tips_df[(tips_df['sex'] == 'Female') & (tips_df['Total_pay'] >= 20)]
```

```python
tips_female_total_pay_df.describe().T
```

```python
tips_male_total_pay_df = tips_df[(tips_df['sex'] == 'Male') | (tips_df['Total_pay'] >= 20)]
```

```python
tips_male_total_pay_df.info()
```

For search several values in feature of DataFrame, as follow:

```python
tips_df[(tips_df['day'].isin(['Sat','Sun'])) & (tips_df['Total_pay'] >= 20) & (tips_df['size'] > 3)]
```
