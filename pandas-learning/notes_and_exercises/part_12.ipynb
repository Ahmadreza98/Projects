```python
import numpy as np
import pandas as pd
```

```python
Identifications_df = pd.DataFrame(
    {
        'ID': [10,20,30,40,50],
        'Name': ['Ali', 'Reza', 'Maryam', 'Mona', 'Nazanin'],
        'BYear': [1395,1375,1355,1358,1366]
    }
)

Bank_info_df = pd.DataFrame(
    {
        'ID': [10,30,40],
        'Balance': [1000,5000,3000]
    }
)
```

Merge two DataFrame

1. Mode:`inner`

```python
pd.merge(left= Identifications_df, right=Bank_info_df, how='inner')
```

2. Mode: `left`

```python
pd.merge(left= Identifications_df, right=Bank_info_df, how='left')
```

3. Mode:`right`

```python
pd.merge(left= Identifications_df, right=Bank_info_df, how='right')
```

4. Mode:`outer`

```python
pd.merge(left= Identifications_df, right=Bank_info_df, how='outer')
```
