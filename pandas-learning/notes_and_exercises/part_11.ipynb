Concatenate two or more DataFrame

```python
df1 = pd.DataFrame(
    {
        'A': ['A0', 'A1', 'A2', 'A3'],
        'B': ['B0', 'B1', 'B2', 'B3'],
        'C': ['C0', 'C1', 'C2', 'C3'],
        'D': ['D0', 'D1', 'D2', 'D3']
    },
    index =[0,1,2,3]
)

df2 = pd.DataFrame(
    {
        'A': ['E0', 'E1', 'E2', 'E3'],
        'B': ['F0', 'F1', 'F2', 'F3'],
        'C': ['G0', 'G1', 'G2', 'G3'],
        'D': ['H0', 'H1', 'H2', 'H3']
    },
    index =[4,5,6,7]
)

df3 = pd.DataFrame(
    {
        'A': ['I0', 'I1', 'I2', 'I3'],
        'B': ['J0', 'J1', 'J2', 'J3'],
        'C': ['K0', 'K1', 'K2', 'K3'],
        'D': ['L0', 'L1', 'L2', 'L3']
    },
    index =[8,9,10,11]
)
```

```python
df_list = [df1, df2, df3]
```

```python
pd.concat(df_list)
```

```python
pd.concat(df_list, axis=1)
```
