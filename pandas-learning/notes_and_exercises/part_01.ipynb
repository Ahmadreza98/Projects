# How to use Pandas?
## part - 1

```python
import pandas as pd
```

**Series** data in pandas

```python
my_data = pd.Series([10,20,30,40], index=["age1", "age2", "age3", "age4"])
```

```python
print(my_data)
```

If we don't define index for data, set numerical default for them.

```python
my_data = pd.Series([10,20,30,40])
```

```python
print(my_data)
```

If my data is dictionary, the Series function set default the key and values are index and data, respectively.

```python
my_dic = {'ali':10, 'mam':20, 'fat':40, 'cat':8}
```

```python
my_data_3 = pd.Series(my_dic)
```

```python
print(my_data_3)
```

Conditional orders can be applied directly to Series, the output is an Series of true and false.

```python
my_data_3[my_data_3 > 20]
```
