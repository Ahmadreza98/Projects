```python
import numpy as np
import pandas as pd
```

```python
tips_df = pd.read_csv("DATA/tips.csv")
```

```python
tips_df.set_index("Payment ID", inplace=True)
```

```python
tips_df.head()
```

**Replace** the **Dinner** to **D** in **DataFrame**

```python
tips_df.replace(to_replace='Dinner', value='D', inplace=True)
```

**Apply** a function with single input on feature of **DataFrame**

```python
tips_df['CC Number'] = tips_df['CC Number'].apply(lambda x: f'{str(x)[:4]} . . . {str(x)[-4:]}')
```

```python
tips_df.head()
```

**Apply** a function with **Multi** input on feature of **DataFrame**

```python
def sum_pay(num1,num2):
    return num1+num2
```

```python
tips_df['total_payment'] = tips_df.apply(lambda row: sum_pay(row['total_bill'], row['tip']), axis=1)
```

```python
tips_df.head()
```

```python
def tip_calc(bill,tip):
    if tip/bill * 100 >= 25:
        return True
    else:
        return False
```

```python
tips_df['generosity_flag'] = tips_df.apply(lambda row: tip_calc(row['total_bill'], row['tip']), axis=1)
```

```python
tips_df[tips_df['generosity_flag'] == True].head()
```

```python
tips_df.drop('generosity_flag', axis=1, inplace=True)
```

**Other Method**

```python
tips_df['generosity_flag'] = np.vectorize(tip_calc)(tips_df['total_bill'],tips_df['tip'])
```
