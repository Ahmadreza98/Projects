```python
import numpy as np
import pandas as pd
```

```python
sales_df = pd.read_csv('DATA/Sales_Funnel_CRM.csv')
```

```python
sales_df.head()
```

The **Pivot** function is a famous, because this function have many attributes for modify **DataFrame**

```python
pd.pivot(sales_df, index='Company', columns='Product', values='Licenses')
```

```python
pd.pivot_table(sales_df, values=['Sale Price', 'Licenses'], index='Company', aggfunc='sum')
```

```python
pd.pivot_table(sales_df, index=['Account Manager', 'Contact'], values='Sale Price', aggfunc='sum', columns='Product', fill_value=0)
```

```python
pd.pivot_table(sales_df, index=['Account Manager', 'Contact', 'Product'], values='Sale Price', aggfunc=[np.sum, np.mean], fill_value=0)
```
