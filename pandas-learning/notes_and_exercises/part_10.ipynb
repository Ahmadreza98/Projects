```python
import numpy as np
import pandas as pd
```

```python
mpg_df.head()
```

One of the popular function in Pandas, `groupby` function.

```python
mpg_df.groupby('model_year').mean(numeric_only=True)
```

```python
mpg_df.groupby('model_year').describe().T
```

```python
mpg_gby_df = mpg_df.groupby(['model_year', 'cylinders']).mean(numeric_only=True)
```

```python
mpg_gby_df.columns
```
