```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Student Name': ['Ali', 'Maryam', 'Reza', 'Sara', 'Amir', 'Zahra', 'Hossein', 'Fatemeh'],
    'Subject': ['Math', 'Physics', 'Chemistry', 'Math', 'Physics', 'Chemistry', 'Math', 'Physics'],
    'Grade': [18, 15, 19, 17, 14, 16, 20, 13],
    'Status': ['Pass', 'Pass', 'Pass', 'Pass', 'Pass', 'Pass', 'Pass', 'Pass'],
    'Sex': ["Male","Female","Male","Female","Male", "Female", "Male","Female"]
}

df = pd.DataFrame(data)
```

```python
df
```

To **Replace** of the values in **DataFrame**, the method is better to other methods.

```python
dic = {'Female':'F', 'Male':'M'}
df.replace(dic,inplace=True)
```

To **show** the first five rows of **DataFrame**, Use the **HEAD** Function

```python
df.head()
```

To **show** the last five rows of **DataFrame**, Use the **HEAD** Function

```python
df.tail()
```

To **Sort** values in **DataFrame**

If `axis=0` than sort by columns.

If `axis=1` than sort by rows.

```python
df.sort_values('Sex', axis=0)
```
